```python
# Program to show how to add, update and remove items in a List, Tuple and Dictionary in Python

# List
numbers = [10, 20, 30]
print("Starting list:", numbers)

# Adding items
numbers.append(40)          # add at the end

# Updating an item
numbers[2] = 25             # change value at index 2

# Removing items
numbers.remove(10)          # remove by value

# TUPLE
data = (1, 2, 3)
print("Starting tuple:", data)

# Adding 
data = data + (4,)

# Updating 
temp = list(data)
temp[1] = 20
data = tuple(temp)

# Removing 
temp = list(data)
temp.remove(3)
data = tuple(temp)


# DICTIONARY 
info = {"a": 1, "b": 2}
print("Starting dictionary:", info)

# Adding 
info["c"] = 3

# Updating a value
info["b"] = 20

# Removing a key
removed = info.pop("a")

# Modified List,Tuple and Dictionary

print("modified list:", numbers)

print("modified tuple:", data)

print("modified dictionary:", info)
```
